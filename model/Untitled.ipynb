```python
print(df.head())
```

```python
df['Ram'] = df['Ram'].str.replace('GB', '')
df['Ram'] = df['Ram'].astype(int)

print(df['Ram'].head())
```

```python
df['Weight'] = df['Weight'].str.replace('kg', '')
df['Weight'] = df['Weight'].astype(float)

print(df['Weight'].head())
```

```python
df['Cpu'] = df['Cpu'].str.extract(
    r'(i3|i5|i7|i9|Ryzen 3|Ryzen 5|Ryzen 7|Ryzen 9|Celeron|Pentium)'
)

df['Cpu'] = df['Cpu'].fillna('Other')

print(df['Cpu'].head())
```

```python
print(df['Memory'].dtype)
```

```python
print(df['Memory'].head())
```

```python
df['Gpu'] = df['Gpu'].str.split().str[0]

print(df['Gpu'].head())
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

kolom_text = [
    'Company',
    'TypeName',
    'Cpu',
    'Gpu'
]

for col in kolom_text:
    df[col] = encoder.fit_transform(df[col])

print(df.head())
```

```python
def kategori_laptop(row):

    if (
        row['Ram'] >= 16
        and row['Gpu'] in ['Nvidia', 'AMD']
    ):
        return 'Gaming'

    elif row['Ram'] >= 8:
        return 'Programming'

    else:
        return 'Office'


df['Kategori'] = df.apply(
    kategori_laptop,
    axis=1
)
```

```python
X = df[
    [
        'TypeName',
        'Ram',
        'Memory',
        'Cpu',
        'Gpu',
        'Weight',
        'Price_euros'
    ]
]

y = df['Kategori']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

best_k = 1
best_acc = 0

for k in range(3, 16):

    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)

    pred = knn.predict(X_test)

    acc = accuracy_score(y_test, pred)

    print(f"K={k}, Accuracy={acc*100:.2f}%")

    if acc > best_acc:
        best_acc = acc
        best_k = k

print("K terbaik:", best_k)
print("Akurasi terbaik:", best_acc)
```
